```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
drug_schema =  StructType([
    StructField("id", StringType(), True),
    StructField("first_name", StringType(), True),
    StructField("last_name", StringType(), True),
    StructField("email", StringType(), True),
    StructField("gender", StringType(), True),
    StructField("ip_address", StringType(), True)
])
```

```python
df_br = spark.readStream.format("cloudFiles") \
    .option("header",'true')\
    .option("cloudFiles.format", "csv") \
    .schema(drug_schema) \
    .load('abfss://rawdata@drugstorageacc.dfs.core.windows.net/')
```

```python
display(df_br)
```

```python
check_loc = 'abfss://silver@drugstorageacc.dfs.core.windows.net/checkpoints'
```

```python
df_br.writeStream.format('delta')\
    .option('checkpointLocation',check_loc)\
    .outputMode('append')\
    .trigger(processingTime="10 seconds")\
    .start('abfss://bronze@drugstorageacc.dfs.core.windows.net/data')
```
