```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
drug_schema =  StructType([
    StructField("_c0", StringType(), True),
    StructField("_c1", StringType(), True),
    StructField("_c2", StringType(), True),
    StructField("_c3", StringType(), True),
    StructField("_c4", StringType(), True),
    StructField("_c5", StringType(), True)
])
```

```python
df_rs = spark.readStream.format("cloudFiles") \
    .option("cloudFiles.format", "csv") \
    .schema(drug_schema) \
    .load('abfss://rawdata@drugstorageacc.dfs.core.windows.net/')
```

```python
display(df_rs)
```

```python
check_loc = 'abfss://silver@drugstorageacc.dfs.core.windows.net/checpoints'
```

```python
df_rs.writeStream.format('csv')\
    .option('checkpointLocation',check_loc)\
    .trigger(processingTime="10 seconds")\
    .start('abfss://bronze@drugstorageacc.dfs.core.windows.net/csvdata')
```
